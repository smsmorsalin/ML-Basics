```python
import numpy as np
```

```python
a = np.random.randint(1,50, (3,4))
print(a)
print()

#convert into 1D array
print(a.ravel())
```

```python
#reshape
c= a.reshape(6,2)
print(c)
```

```python
#vstack()

b = np.random.randint(1,50, (4,3))
d = b.reshape(6,2)
v = np.vstack((c,d) )
```

```python
#hstack()

b = np.random.randint(1,50, (4,3))
d = b.reshape(6,2)
h = np.hstack((c,d) )
h
```

axis = 0 --> rows are added --> vertical stack

axis = 1 --> columns are added --> horizontal stack

```python
a = np.array([[1,2],
             [2,4]])

b = np.array([[3,2],
             [5,4]])

c = np.concatenate((a,b), axis = 0)
print(c)

print()

#transpose matrix
t = c.T
print(t)
```

#basic calculation using np

```python
x = np.array([[1,3],[2,3]])
y = np.array([[1,4],[1,2]])

print("add -->")
print(x+y)
print()

print("sub -->")
print(x-y)
print()

print("mul --> ")
print(x*y)
print()

print("div -->")
print(x/y)
print()

print("power -->")
print(x**2)
print()

print("root -->")
print(x**1/2)
print()

print()
print("using function/method of numpy --->")
print()

print(np.add(x,y))
print()
print(np.subtract(x,y))
print()
print(np.multiply(x,y))
print()
print(np.divide(x,y))
print()
print(np.power(x,2))
print()
print(np.sqrt(x))
```

```python
#range array
a = np.arange(2,10)
print(a)

print(np.exp(a))
print(np.log(a))
print(np.log10(a))
print(np.sin(a))
print(np.cos(a))
print(np.tan(a))
```

```python
#sum, max, min

x = np.array([[1,2], [2,3]])
y = np.array([[1,4], [1,2]])

print("SUM-->", x.sum())
print("MAX-->", x.max())
print("MIN -->",x.min())

#for every row if i want to get the summation
print(np.sum(np.sum(x, axis = 1)))
print(np.max(np.max(x, axis = 1)))
print(np.min(np.min(x, axis = 1)))

print(np.dot(x,y))
```
